# Stress Dataset Generator 
 generates a synthetic correlated stress-management dataset until the CSV reaches approximately 12 GB.

```python
import pandas as pd
import numpy as np
import os
from datetime import datetime

OUTPUT_FILE = "Stress_BigData_12GB.csv"
TARGET_SIZE_GB = 12
BATCH_SIZE = 100_000

rng = np.random.default_rng(42)

occupations = ["Student","Engineer","Doctor","Teacher","Nurse","Business","Driver","Lawyer","Farmer","Police"]
genders = ["Male","Female"]
education = ["School","Diploma","Bachelor","Master","PhD"]
marital = ["Single","Married","Divorced"]
diet = ["Healthy","Average","Poor"]
weather = ["Sunny","Rainy","Cloudy","Hot","Cold"]

if os.path.exists(OUTPUT_FILE):
    os.remove(OUTPUT_FILE)

header = True
user_id = 1

while True:
    age = rng.integers(18,70,BATCH_SIZE)
    sleep = np.clip(rng.normal(6.8,1.5,BATCH_SIZE),2,10)
    workload = rng.integers(1,11,BATCH_SIZE)
    exercise = np.clip(rng.normal(35,20,BATCH_SIZE),0,120)
    financial = rng.integers(1,11,BATCH_SIZE)
    family = rng.integers(1,11,BATCH_SIZE)
    relationship = rng.integers(1,11,BATCH_SIZE)
    social = np.clip(rng.normal(3,2,BATCH_SIZE),0,10)
    mood = np.clip(10-(0.4*workload+0.2*financial)+rng.normal(0,1,BATCH_SIZE),1,10)

    stress = (
        25
        + workload*4
        + financial*2
        + family*1.8
        + relationship*1.5
        + (8-sleep)*5
        - exercise*0.08
        - social*0.8
        - mood*2
        + rng.normal(0,4,BATCH_SIZE)
    )
    stress = np.clip(stress,1,100)

    heart = np.clip(60 + stress*0.6 + rng.normal(0,5,BATCH_SIZE),55,180)
    bp = np.clip(90 + stress*0.7 + rng.normal(0,6,BATCH_SIZE),90,190)
    hrv = np.clip(120 - stress + rng.normal(0,8,BATCH_SIZE),20,120)

    level = np.where(stress<35,"Low",np.where(stress<70,"Medium","High"))

    df = pd.DataFrame({
        "User_ID": np.arange(user_id,user_id+BATCH_SIZE),
        "Age": age,
        "Gender": rng.choice(genders,BATCH_SIZE),
        "Occupation": rng.choice(occupations,BATCH_SIZE),
        "Education": rng.choice(education,BATCH_SIZE),
        "Marital_Status": rng.choice(marital,BATCH_SIZE),
        "Sleep_Hours": np.round(sleep,2),
        "Working_Hours": rng.integers(4,15,BATCH_SIZE),
        "Exercise_Minutes": np.round(exercise,1),
        "Screen_Time_Hours": np.round(rng.uniform(1,12,BATCH_SIZE),2),
        "Social_Media_Hours": np.round(rng.uniform(0,8,BATCH_SIZE),2),
        "Financial_Stress": financial,
        "Family_Stress": family,
        "Relationship_Stress": relationship,
        "Workload": workload,
        "Mood_Score": np.round(mood,2),
        "Heart_Rate": np.round(heart,1),
        "Blood_Pressure": np.round(bp,1),
        "HRV": np.round(hrv,1),
        "BMI": np.round(rng.uniform(18,35,BATCH_SIZE),2),
        "Diet_Quality": rng.choice(diet,BATCH_SIZE),
        "Weather": rng.choice(weather,BATCH_SIZE),
        "Stress_Score": np.round(stress,2),
        "Stress_Level": level,
        "Timestamp": pd.date_range(datetime.now(), periods=BATCH_SIZE, freq="s")
    })

    df.to_csv(OUTPUT_FILE,mode="a",index=False,header=header)
    header=False
    user_id += BATCH_SIZE

    size = os.path.getsize(OUTPUT_FILE)/(1024**3)
    print(f"Rows: {user_id-1:,} | Size: {size:.2f} GB")

    if size >= TARGET_SIZE_GB:
        print("Target size reached.")
        break

print("Done:", OUTPUT_FILE)
```
