```python
import pandas as pd
import numpy as np
import random
import string
import os

# Output file
output_file = "stress_bigdata_7GB.csv"

# Number of rows written in one batch
batch_size = 500000

# Total rows (Adjust if needed)
total_rows = 50000000

# Activity labels
activities = [
    "Working", "Sleeping", "Exercise", "Walking",
    "Studying", "Driving", "Meditation"
]

stress_levels = ["Low", "Medium", "High"]

locations = [
    "Home", "Office", "College",
    "Gym", "Outdoor", "Hospital"
]

def random_text(length=50):
    return ''.join(random.choices(string.ascii_letters + ' ', k=length))

# Remove existing file
if os.path.exists(output_file):
    os.remove(output_file)

written = 0

for start in range(0, total_rows, batch_size):

    current_batch = min(batch_size, total_rows - start)

    df = pd.DataFrame({

        "User_ID": np.arange(start, start + current_batch),

        "Age": np.random.randint(18, 70, current_batch),

        "Gender": np.random.choice(["Male", "Female"], current_batch),

        "Heart_Rate": np.random.randint(55, 180, current_batch),

        "Blood_Pressure": np.random.randint(90, 180, current_batch),

        "Sleep_Hours": np.round(np.random.uniform(3, 10, current_batch),2),

        "Stress_Score": np.random.randint(1, 101, current_batch),

        "Activity": np.random.choice(activities, current_batch),

        "Location": np.random.choice(locations, current_batch),

        "Stress_Level": np.random.choice(stress_levels, current_batch),

        "Temperature": np.round(np.random.uniform(20,40,current_batch),2),

        "Humidity": np.round(np.random.uniform(30,95,current_batch),2),

        "Comment":[random_text(60) for _ in range(current_batch)],

        "Timestamp": pd.date_range(
            "2023-01-01",
            periods=current_batch,
            freq="s"
        )

    })

    df.to_csv(
        output_file,
        mode='a',
        index=False,
        header=not os.path.exists(output_file)
    )

    written += current_batch

    size = os.path.getsize(output_file)/(1024**3)

    print(f"Rows Written: {written:,}")
    print(f"Current Size: {size:.2f} GB\n")

    if size >= 7:
        print("Target size (7 GB) reached.")
        break

print("Dataset Generation Completed.")
```

```python
import pandas as pd
import numpy as np
import random
import os
from datetime import datetime

# -------------------------------
# Configuration
# -------------------------------
OUTPUT_FILE = "StressManagement_BigData.csv"
TARGET_SIZE_GB = 5
BATCH_SIZE = 100000

# -------------------------------
# Categories
# -------------------------------
gender = ["Male", "Female"]

occupation = [
    "Student","Engineer","Doctor","Teacher","Nurse",
    "Business","Lawyer","Police","Driver","Farmer"
]

activity = [
    "Walking","Running","Working","Studying",
    "Sleeping","Exercise","Gaming","Meditation"
]

diet = [
    "Healthy","Average","Poor"
]

smoking = ["Yes","No"]

alcohol = ["Yes","No"]

marital = [
    "Single","Married","Divorced"
]

education = [
    "School","Diploma","Bachelor",
    "Master","PhD"
]

stress_level = [
    "Low","Medium","High"
]

cities = [
    "Delhi","Mumbai","Bangalore","Chennai",
    "Hyderabad","Kochi","Pune","Kolkata"
]

weather = [
    "Sunny","Rainy","Cloudy","Hot","Cold"
]

# -------------------------------
# Long random text
# -------------------------------
comments = [
    "Feeling stressed because of workload.",
    "Good sleep improved my mood.",
    "Traffic increased stress.",
    "Heavy rain affected mental health.",
    "Meditation helped reduce anxiety.",
    "Family support reduced stress.",
    "Exams caused high pressure.",
    "Office deadline created anxiety.",
    "Exercise improved overall wellbeing.",
    "Poor diet affected energy level."
]

# Delete existing file
if os.path.exists(OUTPUT_FILE):
    os.remove(OUTPUT_FILE)

header = True
row_start = 0

while True:

    ids = np.arange(row_start, row_start+BATCH_SIZE)

    df = pd.DataFrame({

        "User_ID": ids,

        "Age": np.random.randint(18,70,BATCH_SIZE),

        "Gender": np.random.choice(gender,BATCH_SIZE),

        "Occupation": np.random.choice(occupation,BATCH_SIZE),

        "Working_Hours": np.random.randint(4,16,BATCH_SIZE),

        "Sleep_Hours": np.round(np.random.uniform(2,10,BATCH_SIZE),2),

        "Heart_Rate": np.random.randint(55,170,BATCH_SIZE),

        "Blood_Pressure": np.random.randint(90,180,BATCH_SIZE),

        "Respiration_Rate": np.random.randint(12,30,BATCH_SIZE),

        "Body_Temperature": np.round(np.random.uniform(35,40,BATCH_SIZE),2),

        "BMI": np.round(np.random.uniform(15,40,BATCH_SIZE),2),

        "Exercise_Minutes": np.random.randint(0,180,BATCH_SIZE),

        "Water_Intake_Liters": np.round(np.random.uniform(0.5,5,BATCH_SIZE),2),

        "Screen_Time_Hours": np.round(np.random.uniform(0,15,BATCH_SIZE),2),

        "Social_Media_Hours": np.round(np.random.uniform(0,12,BATCH_SIZE),2),

        "Study_Hours": np.round(np.random.uniform(0,12,BATCH_SIZE),2),

        "Workload_Score": np.random.randint(1,11,BATCH_SIZE),

        "Financial_Stress": np.random.randint(1,11,BATCH_SIZE),

        "Family_Stress": np.random.randint(1,11,BATCH_SIZE),

        "Relationship_Stress": np.random.randint(1,11,BATCH_SIZE),

        "Traffic_Stress": np.random.randint(1,11,BATCH_SIZE),

        "Academic_Stress": np.random.randint(1,11,BATCH_SIZE),

        "Diet_Quality": np.random.choice(diet,BATCH_SIZE),

        "Smoking": np.random.choice(smoking,BATCH_SIZE),

        "Alcohol": np.random.choice(alcohol,BATCH_SIZE),

        "Meditation_Minutes": np.random.randint(0,120,BATCH_SIZE),

        "Mood_Score": np.random.randint(1,11,BATCH_SIZE),

        "Happiness_Index": np.random.randint(1,101,BATCH_SIZE),

        "Weather": np.random.choice(weather,BATCH_SIZE),

        "City": np.random.choice(cities,BATCH_SIZE),

        "Marital_Status": np.random.choice(marital,BATCH_SIZE),

        "Education": np.random.choice(education,BATCH_SIZE),

        "Stress_Score": np.random.randint(1,101,BATCH_SIZE),

        "Stress_Level": np.random.choice(stress_level,BATCH_SIZE),

        "Activity": np.random.choice(activity,BATCH_SIZE),

        "Comment": np.random.choice(comments,BATCH_SIZE),

        "Timestamp": pd.date_range(
            start=datetime.now(),
            periods=BATCH_SIZE,
            freq="s"
        )

    })

    df.to_csv(
        OUTPUT_FILE,
        mode="a",
        index=False,
        header=header
    )

    header = False
    row_start += BATCH_SIZE

    size = os.path.getsize(OUTPUT_FILE)/(1024**3)

    print(f"Rows Generated : {row_start:,}")
    print(f"Current Size   : {size:.2f} GB")

    if size >= TARGET_SIZE_GB:
        print("\nTarget size reached.")
        break

print("\nDataset successfully created.")
```
